```python
# Task 1

def division(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        print("Zero nahi kar sakte")
    except ValueError:
        print("input nahi diya")
    except Exception as e:
        print("pata nahi kya hua")
```

```python
division(4, 0)
```

```python
division('a', 0)
```

```python
division(5, 2)
```

```python
# Task 2

def sos(n):
    total = 0
    for i in range(n):
        try:
            num = int(input("Enter your value: "))
            total += num ** 2
        except ValueError:
            raise ValueError("Galat input dala")
    return total

def main():
    try:
        n = int(input('Enter total number of values: '))
        result = sos(n)
        print(f"{result}")
    except ValeError:
        print("Galat")

if __name__ == "__main__":
    main()
```

```python
# Task 4

def main():
    try:
        my_list = [1, 2, 3, 4, 5]
        
        index = int(input("Enter the index: "))
        value = my_list[index]
        
        print(value)
        
    except IndexError:
        print("Range ke bhar h")
        
if __name__ == "__main__":
    main()
```
