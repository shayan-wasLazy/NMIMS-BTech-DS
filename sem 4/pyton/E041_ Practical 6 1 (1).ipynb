```python
import F

choice = input("Enter operation sin, cos, sqrt, pow: ")
value = float(input("Enter a value: "))

if choice == 'sin':
    r = F.calc_sin(value)
elif choice == 'cos':
    r = F.calc_cos(value)
elif choice == 'sqrt':
    r = F.calc_sqr(value)
elif choice == 'pow':
    exp = int(input("Enter exponent: "))
    r = F.calc_pow(value, exp)
else:
    print("Invalid Choice!!")

    
print(r)
```
