```python
# Pattern 1

n = 5

for i in range(1, n+1):
    for k in range(1, i+1):
        print("*", end="")
    print()
```

```python
# Pattern 2

n = 5
for i in range(n):
    for j in range(1, n-i):
        print(" ", end="")
    for k in range(0, i +1):
        print("*", end="")
    print()
    
```

```python
# Pattern 3

n = 5

for i in range(n):
    for j in range(n - i):
        print("*", end="")
    print()
```

```python
# patter 4

for i in range(n):
    for j in range(i):
        print(" ", end="")
    for j in range(n, i, -1):
        print("*", end="")
    print()
```

```python
# Task 6 
data = [{'item': 'item1', 'amount': 400}, {'item': 'item2', 'amount': 300}, {'item': 'item1', 'amount': 750}]

cval = {}
for d in data:
    item = d["item"]
    amount = d["amount"]
    
    if item in cval:
        cval[item] += amount
    else:
        cval[item] = amount
    
print(cval)
```

```python
# Task 7

shop_items = {'item1': 45.50, 'item2':35, 'item3': 41.30, 'item4':55, 'item5': 24}

t3i = []
for i in range(3):
    max_item = max(shop_items, key=shop_items.get)
    t3i.append((max_item, shop_items.pop(max_item)))
    
print("Expected output: ")
for item, price in t3i:
    print(f"{item} {price}")
```

```python
# Task 8

student_dict = {'Cierra Vega': (6.2, 70), 'Alden Cantrell': (5.9, 65), 'Kierra Gentry': (6.0, 68), 'Pierre Cox': (5.8, 66)}

a = float(input("Enter min height: "))
b = float(input("Enter max height: "))
c = float(input("Enter min weight: "))
d = float(input("Enter max weight: "))

min_h = a
max_h = b
min_w = c
max_w = d

for name, (height, weight) in student_dict.items():
    if min_h <= height <= max_h and min_w <= weight <= max_w:
        print(f"{name}: Height - {height} ft, weight - {weight} lbs")
```
