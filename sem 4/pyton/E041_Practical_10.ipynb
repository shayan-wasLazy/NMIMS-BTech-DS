```python
# Task 6

import re

string = "22-09-2024"
r = re.search(r"(\d{2})-(\d{2})-(\d{4})", string)
if r:
    day = r.group(1)
    month = r.group(2)
    year = r.group(3)
    print(day)
    print(month)
    print(year)
```
