```python
# Pattern 1

n = 5

for i in range(1, n+1):
    for k in range(1, i+1):
        print("*", end="")
    print()
```

```python
# Pattern 2

n = 5
for i in range(n):
    for j in range(1, n-i):
        print(" ", end="")
    for k in range(0, i +1):
        print("*", end="")
    print()
    
```

```python
# Pattern 3

n = 5

for i in range(n):
    for j in range(n - i):
        print("*", end="")
    print()
```

```python
# patter 4

for i in range(n):
    for j in range(i):
        print(" ", end="")
    for j in range(n, i, -1):
        print("*", end="")
    print()
```

```python
# Task 6 
data = [{'item': 'item1', 'amount': 400}, {'item': 'item2', 'amount': 300}, {'item': 'item1', 'amount': 750}]

cval = {}
for d in data:
    item = d["item"]
    amount = d["amount"]
    
    if item in cval:
        cval[item] += amount
    else:
        cval[item] = amount
    
print(cval)
```
