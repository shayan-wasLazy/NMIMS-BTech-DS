```python
import tkinter as tk

def ds():
    so = []
    if op1.get() == 1:
        so.append("male")
    if op2.get() == 1:
        so.append("female")
    if op3.get() == 1:
        so.append("non-binary")
        
    print(so)
    
root = tk.Tk()

op1 = tk.IntVar()
cb1 = tk.Checkbutton(root, text = "Option 1", variable = op1, command = ds)
cb1.pack()

op2 = tk.IntVar()
cb2 = tk.Checkbutton(root, text = "Option 2", variable = op2, command = ds)
cb2.pack()

op3 = tk.IntVar()
cb3 = tk.Checkbutton(root, text = "Option 3", variable = op3, command = ds)
cb3.pack()

root.mainloop()
```

```python
from tkinter import *
from tkinter import messagebox

root = Tk()
root.title("Hobbies")
root.geometry("200x200")

var1 = IntVar()
var2 = IntVar()
var3 = IntVar()
var4 = IntVar()


chkbtn1 = Checkbutton(root, text = "Valorant", width = 15, onvalue = 1, offvalue = 0, variable = var1)
chkbtn1.pack()

chkbtn2 = Checkbutton(root, text = "CS2", width = 15, onvalue = 1, offvalue = 0, variable = var2)
chkbtn2.pack()

chkbtn3 = Checkbutton(root, text = "Roblox", width = 15, onvalue = 1, offvalue = 0, variable = var3)
chkbtn3.pack()

chkbtn4 = Checkbutton(root, text = "The Finals", width = 15, onvalue = 1, offvalue = 0, variable = var4)
chkbtn4.pack()

root.mainloop()
```

```python
def onClick():
    sc = ""
    if var1.get() != 0:
        sc += "suiiiiiiiii\n"
    if var2.get() != 0:
        sc += "ankara messe\n"
    if var1.get() != 0:
        sc += "hala madrid\n"
        
    messagebox.showinfo("Hmm", "CHoices are: \n" + sc)
    
def clearChoices():
    var1.set(0)
    var2.set(0)
    var3.set(0)
    var4.set(0)
    
btn_submit = Button(root, text = "Submit", command = onClick)
btn_submit.pack()
btn_clear = Button(root, text = "Clear", command = ClearChoices)
btn_clear.pack()
root.mainloop()
```
