```python
#task d
def maxi(number):
    maximum  = 0
    for num in number:
        if maximum < num:
            maximum = num
    return maximum

sample = (1,6,9)
maximum = maxi(sample)
print(f"Maximum number is {maximum}")
        
```

```python
import math

def gcd_function(a, b):
    return math.gcd(a, b)

def my_gcd_function(a, b):
    while b:
        a, b = b, a % b
        print(a, b)
    return a

num1 = int(input("Enter first number for GCD "))
num2 = int(input("Enter second number for GCD"))

gcd_result = gcd_function(num1, num2)
print(f"GCD using custom function: {gcd_result}")

my_gcd_result = my_gcd_function(num1, num2)
print(f"GCD using custom function: {my_gcd_result}")
```

```python
def fact(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * fact(n - 1)

result = fact(5)
print(result)
```
