```python
def simple_reflex(location, status):
    """
    Simple Reflex Agent
    Makes decisions only based on the current input.
    It has no memory.
    """

    if status == "Dirty":
        return "Suck"

    elif location == "A":
        return "Move Right"

    else:
        return "Move Left"

print(" SIMPLE REFLEX AGENT ")
location = "A"
status = "Dirty"
action = simple_reflex(location, status)
print("Current Location :", location)
print("Room Status      :", status)
print("Action           :", action)

print("\n\n MODEL-BASED REFLEX AGENT ")

environment = {
    "A": "Dirty",
    "B": "Dirty"
}

def model_based_agent(location):
    if environment[location] == "Dirty":
        environment[location] = "Clean"
        return "Suck"
    elif location == "A":
        return "Move Right"
    else:
        return "Move Left"

current = "A"

for step in range(1, 7):
    action = model_based_agent(current)
    print("\n                   ")
    print("Step :", step)
    print("Current Room :", current)
    print("Action :", action)
    print("Environment :", environment)

    if action == "Move Right":
        current = "B"
    elif action == "Move Left":
        current = "A"
```
