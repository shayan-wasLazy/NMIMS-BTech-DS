```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import statsmodels.graphics.tsaplots as sgt
import statsmodels.tsa.stattools as sts
from statsmodels.tsa.seasonal import seasonal_decompose

import seaborn as sns
sns.set()

print("Libraries imported")
```

```python
df_index = pd.read_csv(r"data/Index2018.csv")
```

```python
df_index.info()
```

```python
df_index.head()
```

```python
df_comp = df_index.copy()
df_comp.date = pd.to_datetime(df_comp.date, dayfirst=True)
```

```python
df_comp.info()
```

```python
df_comp.set_index("date", inplace=True)
```

```python
df_comp.head()
```

```python
df_comp = df_comp.asfreq('b')
df_comp = df_comp.bfill()
```

```python
df_comp["market_value"] = df_comp["spx"]
del df_comp["spx"]
del df_comp["dax"]
del df_comp["ftse"]
del df_comp["nikkei"]
```

```python
size = int(len(df_comp) * 0.8)
df_train, df_test = df_comp.iloc[:size] , df_comp.iloc[size:]
df_train.shape
```

```python
wn = np.random.normal(loc = df_train.market_value.mean(), scale = df_train.market_value.std(), size=len(df_train))
```

```python
plt.plot(wn)
plt.show()
```

```python
df_train['wn'] = wn
```

```python
df_train
```

```python
df_train.wn.plot(figsize = (20, 5))
plt.show()
```

```python
df_train.market_value.plot(figsize = (20, 5))
plt.show()
```

```python
rw = pd.read_csv(r"data/RandWalk.csv")
rw.date = pd.to_datetime(rw.date, dayfirst=True)
rw.set_index("date", inplace=True)
rw = rw.asfreq("b")
```

```python
rw.describe()
```

```python
df_train["rw"] = rw
```

```python
df_train.rw.plot(figsize = (20, 5))
df_train.market_value.plot()
plt.show()
```

```python
# 0.5 < non - stationary
# 0.5 > stationary

sts.adfuller(df_train.market_value)
```

```python
sts.adfuller(df_train.wn)
```

```python
sts.adfuller(df_train.rw)
```

```python
s_dec_multiplicative = seasonal_decompose(df_train.market_value,
                                         period=1,
                                         model="multiplicative")
s_dec_multiplicative.plot()
plt.show()
```

```python
sgt .plot_acf(df_train.market_value, lags=15, zero=False)
plt.title("ACF S&P", size = 24)
plt.show()
```
