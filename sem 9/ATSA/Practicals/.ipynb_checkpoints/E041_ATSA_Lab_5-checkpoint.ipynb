````python
import pandas as pd
import numpy as np
import statsmodels.graphics.tsaplots as sgt
import matplotlib.pyplot as plt
````

````python
df = pd.read_csv("AirPassengers.csv")
df["date"] = pd.to_datetime(df["date"])
df = df.set_index("date")
df = df.sort_index()
df.head()
````

````python
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(df.index, df["value"], marker="o", linestyle="-")
ax.set_title("Time Series Data Over Time")
ax.set_xlabel("Date")
ax.set_ylabel("Measured Value")
ax.grid(True)
plt.tight_layout()
plt.show()
````

````python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10))

# ACF plot
sgt.plot_acf(df["value"], ax=ax1, lags=40, alpha=0.05)
ax1.set_title("Autocorrelation Function (ACF)", fontsize=14, 
              fontweight="bold")
ax1.set_xlabel("Lag")
ax1.set_ylabel("Autocorrelation")

# PACF plot
sgt.plot_pacf(df["value"], ax=ax2, lags=40, alpha=0.05)
ax2.set_title("Partial Autocorrelation Function (PACF)", 
              fontsize=14, fontweight="bold")
ax2.set_xlabel("Lag")
ax2.set_ylabel("Partial Autocorrelation")

plt.tight_layout()
plt.show()
````

````python
df_lag = df.copy()
for i in range(1, 5):
    df_lag[f"lag_{i}"] = df["value"].shift(i)
df_lag.head()
df_lag = df_lag.dropna()
df_lag.head()
````

````python
lag_cols = [c for c in df_lag.columns if c.startswith("lag_")]
X = df_lag[lag_cols]
y = df_lag["value"]
````

````python
X.head()
````

````python
y.head()
````

````python
split_idx = int(0.80 * len(df_lag))
X_train = X.iloc[:split_idx].copy()
X_test = X.iloc[split_idx:].copy()
y_train = y.iloc[:split_idx].copy()
y_test = y.iloc[split_idx:].copy()
````

````python
from sklearn.linear_model import LinearRegression
````

````python
model = LinearRegression()
model.fit(X_train, y_train)
````

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


y_pred = model.predict(X_test)
pred = pd.Series(y_pred, index=y_test.index, name="predicted")

mae = mean_absolute_error(y_test, pred)
mse = mean_squared_error(y_test, pred)
rmse = np.sqrt(mean_squared_error(y_test, pred))
r2 = r2_score(y_test, pred)
print(f"Test MAE:  {mae:.3f}")
print(f"Test MSE: {mse:.3f}")
print(f"Test RMSE: {rmse:.3f}")
print(f"Test R²:   {r2:.3f}")

# Plot: actual vs predicted (test period)
plt.figure(figsize=(12, 6))
plt.plot(y_test.index, y_test, label="Actual (test)", linewidth=2)
plt.plot(pred.index, pred, label="Predicted", linewidth=2)
plt.title("Test Set: Actual vs Predicted")
plt.xlabel("Date")
plt.ylabel("Passengers")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
````

````python
!pip install torch torchvision torchaudio
import torch
import torch.nn as nn
````

````python
class ANNModel(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super().__init__()
        self.input_lr = nn.Linear(input_size, hidden_size)
        self.hidden_lr = nn.Linear(hidden_size, hidden_size)
        self.output_lr = nn.Linear(hidden_size, output_size)
        self.relu = nn.ReLU()

    def forward(self, x):
        x = self.relu(self.input_lr(x))
        x = self.relu(self.hidden_lr(x))
        x = self.output_lr(x)
        return x
````

````python
# Convert data to PyTorch tensors
X_train_tensor = torch.tensor(X_train.values, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train.values, 
                              dtype=torch.float32).reshape(-1, 1)
X_test_tensor = torch.tensor(X_test.values, 
                             dtype=torch.float32)
y_test_tensor = torch.tensor(y_test.values, 
                             dtype=torch.float32).reshape(-1, 1)

# Initialize model, loss function, and optimizer
input_size = X_train.shape[1]  # 4 lag features
hidden_size = 32
output_size = 1

ann_model = ANNModel(input_size, hidden_size, output_size)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(ann_model.parameters(), lr=0.001)

# Training loop
epochs = 500
train_losses = []
test_losses = []

for epoch in range(epochs):
    ann_model.train()

    # Forward pass
    y_pred_train = ann_model(X_train_tensor)
    loss = criterion(y_pred_train, y_train_tensor)

    # Backward pass
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    # Track losses
    train_losses.append(loss.item())

    # Evaluate on test set
    ann_model.eval()
    with torch.no_grad():
        y_pred_test = ann_model(X_test_tensor)
        test_loss = criterion(y_pred_test, y_test_tensor)
        test_losses.append(test_loss.item())

    if (epoch + 1) % 100 == 0:
        print(
            f"Epoch [{epoch + 1}/{epochs}],Train Loss: {loss.item():.4f}, Test Loss: {test_loss.item():.4f}"
        )

print("Training complete!")
````

````python
plt.figure(figsize=(12, 8))

plt.subplot(2, 1, 1)
plt.plot(train_losses, label="Train Loss")
plt.plot(test_losses, label="Test Loss")
plt.title("Loss Over Epochs")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.legend()
plt.grid(True)

# Get final predictions
ann_model.eval()
with torch.no_grad():
    y_pred_ann = ann_model(X_test_tensor).numpy().flatten()

# Calculate metrics
mae_ann = mean_absolute_error(y_test, y_pred_ann)
mse_ann = mean_squared_error(y_test, y_pred_ann)
rmse_ann = np.sqrt(mean_squared_error(y_test, y_pred_ann))
r2_ann = r2_score(y_test, y_pred_ann)

print(f"\nANN Test MAE:  {mae_ann:.3f}")
print(f"ANN Test MSE:  {mse_ann:.3f}")
print(f"ANN Test RMSE: {rmse_ann:.3f}")
print(f"ANN Test R²:   {r2_ann:.3f}")

# Plot actual vs predicted
plt.subplot(2, 1, 2)
plt.plot(y_test.index, y_test.values, 
         label="Actual", linewidth=2)
plt.plot(y_test.index, y_pred_ann, 
         label="Predicted (ANN)", linewidth=2)
plt.title("Test Set: Actual vs Predicted (ANN)")
plt.xlabel("Date")
plt.ylabel("Passengers")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
````

##### Linear Relationships in the Data
- The AirPassengers data exhibits strong **linear autocorrelation** — past values directly influence future values in a relatively linear manner.
- MLR is designed to capture linear relationships, making it naturally suited for this task.

##### Key Takeaway
> For small, linearly-structured time series data, **simpler models like MLR often outperform complex neural networks**. ANNs shine when there are non-linear patterns and large amounts of training data available.
