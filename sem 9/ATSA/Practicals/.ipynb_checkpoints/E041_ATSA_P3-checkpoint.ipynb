```python
#Pandas for data manipulation
import pandas as pd
#Numpy for numerical operations
import numpy as np
#Matplotlib for plotting
import matplotlib.pyplot as plt
#Statsmodels for time series analysis and plotting
import statsmodels.graphics.tsaplots as sgt
import statsmodels.tsa.stattools as sts
from statsmodels.tsa.seasonal import seasonal_decompose
#Seaborn for enhanced visuals
import seaborn as sns
from statsmodels.graphics.tsaplots import plot_acf,plot_pacf
from pandas.plotting import lag_plot
```

```python
df = pd.read_csv(r"data/AirPassengers.csv")
df['date'] = pd.to_datetime(df['date'])
df.set_index('date',inplace=True)
 
df['year'] = df.index.year
df['month'] = df.index.month
df['month_name'] = df.index.strftime('%b')
 
print("Data Loading done and grouped columns created")
df.head()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df.index, df['value'], color='tab:blue', linewidth=2)
plt.title("Task 1 : Line (Time) Plot of Air Passengers")
plt.xlabel("Year")
plt.ylabel("Number of Passengers")
plt.grid(True,alpha=0.3)
plt.show()
```

### Seasonal Plot

```python
plt.figure(figsize=(12, 6))

for year, group in df.groupby("year"):
    plt.plot(group["month_name"], group["value"], label = year)
plt.title("Task 2: seasonal Plot (Year-wise comaprison using groupby)")
plt.xlabel("Month")
plt.ylabel("Passengers")
plt.legend(bbox_to_anchor = (1.05, 1), loc = "upper left", ncol = 2)
plt.grid(True,alpha=0.3)
plt.show()
```

```python
grouped_months = df.groupby("month")
fig, axes = plt.subplots(3, 4, figsize = (15, 10))
axes = axes.flatten()

for i, (month_num, group) in enumerate(grouped_months):
    ax = axes[i]
    ax.plot(group["year"], group["value"], marker = "o", markersize = 3)
    ax.set_title(group["month_name"].iloc[0])
    ax.tick_params(axis = "x", rotation = 45)

plt.tight_layout()
plt.suptitle("Task 3: Subseries Plot (Monthly growth trends using Groupby)", y = 1.02, fontsize = 16)
plt.show()
```

```python
fig, axes = plt.subplots(3, 4, figsize = (15, 10))
axes = axes.flatten()

for i in range(1, 13):
    lag_plot(df["value"], lag = i, ax = axes[i-1], s = 10, alpha = 0.6)
    axes[i - 1].set_title(f"Lag{i}")

plt.tight_layout()
plt.suptitle("Task 4: Sactter Plots (Lag 1 to 12)", y = 1.02, fontsize = 16)
plt.show()
```

```python
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

fig, (ax1,ax2) = plt.subplots(2,1, figsize=(12,10))
plot_acf(df["value"], lags=40, ax=ax1)
ax1.set_title("Task 5 : Auto - correlation Function ( ACF )")
 
plot_pacf(df["value"], lags=40, ax=ax2)
ax1.set_title("Task 6 : Partial Auto - correlation Function ( PACF )")
```
