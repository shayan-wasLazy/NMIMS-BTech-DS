```python
# 3layer (2 hiden with 4 node and 1 output layer)
import numpy as np

def relu(x):
    return np.maximum(0,x)

def sigmoid(x):
    return 1/(1+np.exp(-x))

# step 1 nodes and layers
def nodes_size(x, y, hidden_nodes):
    nx = x.shape[0]        
    nh1 = hidden_nodes[0]  
    nh2 = hidden_nodes[1]  
    ny = y.shape[0]        
    return (nx, nh1, nh2, ny)
```

```python
def int_parameters(nx, nh1, nh2, ny):
    np.random.seed(123)
    
    w1 = np.random.randn(nh1, nx)
    b1 = np.zeros((nh1, 1))
    
    w2 = np.random.randn(nh2, nh1)
    b2 = np.zeros((nh2, 1))
    
    w3 = np.random.randn(ny, nh2)
    b3 = np.zeros((ny, 1))
    
    parameters = {"W1": w1, "b1": b1, "W2": w2, "b2": b2,  "W3": w3, "b3": b3}
    return parameters

def fwdp(x, params):
    w1 = params["W1"]
    b1 = params["b1"]
    w2 = params["W2"]
    b2 = params["b2"]
    w3 = params["W3"]
    b3 = params["b3"]
    
    z1 = np.dot(w1, x) + b1
    a1 = np.tanh(z1)
    z2 = np.dot(w2, a1) + b2
    a2 = relu(z2)
    z3 = np.dot(w3, a2) + b3
    a3 = sigmoid(z3)
    
    cache = { "z1": z1, "a1": a1, "z2": z2, "a2": a2,"z3": z3, "a3": a3}
    return a3, cache

def compute_cost(a3, y):
    m = y.shape[1]
    cost1 = np.sum(y * np.log(a3) + (1 - y) * np.log(1 - a3))
    cost = -cost1 / m
    cost = float(np.squeeze(cost))
    return cost
```

```python
#5 backward propagation
def BWP(params, cache, X,Y):
    w1 = params["W1"]
    b1 = params["b1"]
    w2 = params["W2"]
    b2 = params["b2"]
    w3 = params["W3"]
    b3 = params["b3"]

    a1 = cache["a1"]
    a2 = cache["a2"]
    a3 = cache["a3"]


    m = Y.size
    dz3 = a3 - Y
    dw3 = np.dot(dz3,a2.T)/m
    dz2 = a2 - Y
    db3 = np.sum(dz3,axis=1, keepdims=True)/m

    dw2 = np.dot(dz2,a1.T)/m
    db2 = np.sum(dz2,axis=1, keepdims=True)/m
    da1 = np.dot(w2.T,dz2)
    dz1 = da1*(1-np.power(a1,2))
    dw1 = np.dot(dz1,X.T)/m
    db1 = np.sum(dz1,axis=1, keepdims=True)/m
    grades = {"dw1":dw1,"db1": db1, "dw2":dw2, "db2": db2, "dw3":dw3, "db3": db3}
    
    return grades
```

```python
def update(params, grades, lr=0.01):
    W1 = params['W1']
    w2 = params['W2']
    w3 = params['W3']

    b1 = params['b1']
    b2 = params['b2']
    b3 = params['b3']


    dw1 = grades["dw1"]
    db1 = grades["db1"]
    dw2 = grades["dw2"]
    db2 = grades["db2"]
    dw3 = grades["dw3"]
    db3 = grades["db3"]
 
 
    W1 = W1 - lr*dw1
    b1 = b1 - lr*db1
    w2 = w2 - lr*dw2
    b2 = b2 - lr*db2
    w3 = w3 - lr*dw3
    b3 = b3 - lr*db3

    params = {'W1':W1,   
              "b1": b1,
            "W2": w2,
            "b2": b2,
            "W3": w3,
            "b3": b3}
    
    return params
```

```python
def NN(x,y,nodes,itr=10000,print_cost=False):
    np.random.seed(3)
    nx, nh1, nh2, ny = nodes_size(x,y,nodes)
    params = int_parameters(nx, nh1, nh2, ny)
    for i in range(0, itr):
        a3,cache = fwdp(x,params)
        cost = compute_cost(a3,y)
        grades = BWP(params, cache, x,y)
        params = update(params, grades, lr= 0.01)

        if print_cost & i % 1000 == 0:
            print("cost % i:%f" %(i,cost))
    return params
```

```python
np.random.seed(1)
x = np.random.randn(2,3)
y = (np.random.randn(1,3)>0)
print(x)
print(y)
NN(x,y,nodes=[4, 4],itr=100000,print_cost=False)
```
