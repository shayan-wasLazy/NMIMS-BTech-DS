```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
def sigmoid_grad(x):
    fx = sigmoid(x)
    return fx - (1 - fx)
```

```python
#nodes and layers
layers = 2

def nodes_size(X, y, nodes):
    
    n_x = X.shape[0]
    n_h = nodes
    n_y = y.shape[0]

    return n_x, n_h, n_y
```

```python
n_x, n_h, n_y = nodes_size(X, y, 4)
```

```python
# 2. initialize parameters


def int_parameters(n_x, n_h, n_y):
    np.random.seed(2)
    w1 = (np.random.randn(n_h, n_x)) * 0.01
    b1 = np.random.randn(n_h, 1)
    w2 = (np.random.randn(n_y, n_h)) * 0.01
    b2 = np.random.randn(n_y , 1)

    parameters = {"w1" : w1, "b1" : b1, "w2": w2, "b2" : b2}
    return parameters
```

```python
params = int_parameters(n_x, n_h, n_y)
```

```python
# 3. forward prop

def fwdp(X, params):
    w1 = (params["w1"])
    w2 = (params["w2"])
    b1 = params["b1"]
    b2 = params["b2"]

    z1 =  w1 @ X + b1

    # funcx = int(input("1. sigmoid, 2. tanh: "))
    # if funcx == 1:
    #     a1 = sigmoid(z1)
    # else:
    a1 = np.tanh(z1)

    z2 = w2 @ a1 + b2
    a2 = sigmoid(z2)

    cache = {"z1" : z1, "a1" : a1, "z2" : z2, "a2" : a2}

    return a2, cache
```

```python
a2, cache= fwdp(X, params)
```

```python
# 4. cost

def compute_cost(a2, y):
    m = y.shape[1]

    a2 = np.clip(a2, 1e-15, 1 - 1e-15)

    cost1 = np.sum(
        y * np.log(a2) +
        (1 - y) * np.log(1 - a2)
    )

    cost = -cost1 / m
    cost = float(np.squeeze(cost))

    return cost
```

```python
def bwp(params, cache, X, y):
    np.random.seed(3)
    w1 = params["w1"]
    w2 = params["w2"]
    b1 = params["b1"]
    b2 = params["b2"]

    a1 = cache["a1"]
    a1 = cache["a2"]

    m = y.size  ## m = y.shape[1]

    dz2 = a2 - y
    dw2 = np.dot(dz2, a1.T) / m
    db2 = np.sum(dz2, axis=1, keepdims=True) / m
    da1 = np.dot(w2.T, dz2)
    dz1 = da1 * (1 - np.power(a1, 2))
    dw1 = np.dot(dz1, X.T) / m
    db1 = np.sum(dz1, axis=1, keepdims=True) / m

    grades = {"dw1" : dw1, "db1" : db1, "dw2" : dw2, "db2" : db2}

    return grades
    
```

```python
grades = bwp(params, cache, X, y)
```

```python
def update(params, grades, lr = 0.001):
    w1 = params["w1"]
    w2 = params["w2"]
    b1 = params["b1"]
    b2 = params["b2"]

    dw1 = grades["dw1"]
    dw2 = grades["dw2"]
    db1 = grades["db1"]
    db2 = grades["db2"]

    w1 = w1 - dw1 * lr
    w2 = w2 - dw2 * lr
    b1 = b1 - db1 * lr
    b2 = b2 - db2 * lr

    params = {"w1" : w1, "b1" : b1, "w2": w2, "b2" : b2}

    return params
```

```python
params = update(params, grades)
```

```python
def nn(X, y, layers = 4, itr = 1000, print_cost = False):
    np.random.seed(3)
    n_x, n_h, n_y = nodes_size(X, y, 4)
    params = int_parameters(n_x, n_h, n_y)
    for i in range(itr):
        a2, cache = fwdp(X, params)
        cost = compute_cost(a2, y)
        grades = bwp(params, cache, X, y)
        params = update(params, grades, lr = 0.001)

        if print_cost and i % 1000 == 0:
            print("cost % i:%f" % (i, cost))
```

```python
nn(X, y, layers, itr=100000,print_cost=True)
```
