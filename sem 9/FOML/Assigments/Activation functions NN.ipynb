```python
import numpy as np
```

# Identity

```python
def identity(x):
    return x

def identity_grad(x):
    return np.ones_like(x)
```

```python
x = np.array([1, 4, 7, 8])
x = identity(x)
identity_grad(x)
```

# Binary Step

```python
def binaryStep(x):
    return np.where(x > 0, 1.0, 0.0)

def binaryStep_grad(x):
    grad = np.zeros_like(x, dtype=float)
    grad[x == 0] = np.nan
    return grad
```

```python
x = np.array([-1, 4, 0, 8])
x = binaryStep(x)
print(x)
print(binaryStep_grad(x))
```

# Sigmoid

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
def sigmoid_grad(x):
    fx = sigmoid(x)
    return fx - (1 - fx)
```

```python
x = np.array([1, 4, 7, 8])
x = sigmoid(x)
print(x)
print(sigmoid_grad(x))
```

# TanH

```python
def tanh(x):
    return np.tanh(x)
def tanh2(x):
    tan = (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))
    return tan

def tanh_grad(x):
    return 1 - np.tanh(x) ** 2
```

```python
x = np.array([1, 4, 7, 8])
x = tanh(x)
print(x)
print(tanh_grad(x))
```

```python
x = np.array([1, 4, 7, 8])
x = tanh2(x)
print(x)
print(tanh_grad(x))
```

# ArchTan

```python
def archtan(x):
    return np.arctan(x)
def archtan_grad(x):
    return 1 / (1 + x ** 2)
```

```python
x = np.array([1, 4, 7, 8])
x = archtan(x)
print(x)
print(archtan_grad(x))
```

# Relu

```python
def relu(x):
    return np.maximum(0, x)
def relu_grad(x):
    return np.where(x > 0, 1.0, 0.0)
```

```python
x = np.array([1, 4, -3, 8])
x = relu(x)
print(x)
print(relu_grad(x))
```

# Prelu

```python
def prelu(x, alpha = 0.01):
    return np.where(x > 0, x, alpha * x)
def prelu_grad(x, alpha = 0.01):
    return np.where(x > 0, 1, alpha)
```

```python
x = np.array([1, 4, -3, 8])
x = prelu(x)
print(x)
print(prelu_grad(x))
```

# Erelu

```python
def erelu(x, alpha = 0.01):
    return np.where(x > 0, x, alpha * np.exp(x) - 1)
def erelu_grad(x, alpha = 0.01):
    fx = erelu(x)
    return np.where(x > 0, 1, fx + alpha)
```

```python
x = np.array([1, 4, -3, 8])
x = erelu(x)
print(x)
print(erelu_grad(x))
```
