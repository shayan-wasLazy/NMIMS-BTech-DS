```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(5)
X = np.array([[5,3],[6,4],[7,6],[9,8]])
Y = np.array([2,4,7,8]).reshape(-1,1)
print(X.shape)
print(Y.shape)
```

```python
def LGD(x,y,lr,epoch):
    nx = x.shape[1]
    m = x.shape[0]
    np.random.seed(123)
    w = np.random.rand(nx,1)
    b = 0
    cost=[]
    for i in range(epoch):
        h = np.dot(x,w) + b
        error = h - y
#   cost1 = (1/2*m) * np.sum(error**2)
        cost1 = (1/(2*m)) * np.dot(error.T,error)
        cost.append(cost1)
        dw = (1/m)*(np.dot(x.T,error))
        db = np.sum(error)/m
        w = w-lr*dw
        b = b-lr*db
    cost = np.asarray(cost).reshape(-1,1)
    plt.plot(cost)
    plt.show()
    return w,b,cost[-1]
    
```

```python
LGD(X,Y,0.01,100)
```
