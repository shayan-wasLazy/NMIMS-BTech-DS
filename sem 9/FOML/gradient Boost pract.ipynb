```python
# input: x, y, t0, t1, epochs

import numpy as np
from tqdm import tqdm

def calcGrad(x, y, t0, t1):
    
    yhat = []
    for i in range(len(x)):
        yh = t0 + x[i] * t1
        yhat.append(yh)
    # print(f"yhat values : {yhat}")
    
    ress = []
    for i in range(len(y)):
        c = (y[i] - yhat[i])
        ress.append(c)
    # print(f"cost values : {loss}")

    resst0 = np.average(ress) 

    # print(f"Loss : {resst0}")

    ress = []
    for i in range(len(y)):
        c = x[i] * (y[i] - yhat[i])
        ress.append(c)
    # print(f"cost values : {loss}")

    resst1 = np.average(ress) 

    # print(f"Loss : {resst1}")

    loss = []
    for i in range(len(y)):
        c = (y[i] - yhat[i]) ** 2
        loss.append(c)
    # print(f"cost values : {loss}")

    eloss = np.average(loss)

    return resst0, resst1, eloss

def diffLoss(resst0, resst1):
    dj0 = 2 * resst0 * -1
    dj1 = -2 * resst1
    return dj0, dj1

def main():
    x = [1, 2, 4, 7, 10]
    y = [3, 10, 14, 9, 15]
    t0 = 0.2
    t1 = 0.8
    epochs = 10
    alpha = 0.01
    loss_history = []
    for i in tqdm(range(epochs)):
        resst0, resst1, eloss = calcGrad(x, y, t0, t1)
        dj0, dj1 = diffLoss(resst0, resst1)
        t0 = t0 - alpha * dj0
        t1 = t1 - alpha * dj1
        loss_history.append(eloss)
        print(f"Epoch {i}")
        print(f"t0 = {t0:.4f}")
        print(f"t1 = {t1:.4f}")
        print(f"Loss = {eloss:.4f}")
        print("-------------------")

        import matplotlib.pyplot as plt

    plt.figure(figsize=(6,4))
    plt.plot(loss_history, marker="o")
    plt.xlabel("Epoch")
    plt.ylabel("MSE Loss")
    plt.title("Gradient Descent Convergence")
    plt.grid(True)
    plt.show()
```

```python
main()
```
