```python
import pandas as pd
df=pd.read_excel('ANOVA.xlsx')
df
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

fit1=ols('Method ~ Var', data=df).fit()
anova1=sm.stats.anova_lm(fit1, typ=1)
anova1
# PR(>F) is the P value which is less than 0.05


```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
tukey=pairwise_tukeyhsd(df["Method"], groups=df["Var"])
tukey._results_table

```

```python
import pandas as pd
df=pd.read_csv('auto.csv')
df
```

```python
#1 USA, 2 Europe, 3Japan
#mileage
from statsmodels.stats.multicomp import pairwise_tukeyhsd
tukey=pairwise_tukeyhsd(df["mpg"], groups=df["origin"])
tukey._results_table #all were different in mileage
```
