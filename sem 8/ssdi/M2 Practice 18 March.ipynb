```python
import numpy as np
from scipy import stats
from statistics import stdev
```

```python
#A pharmaceutical company tests whether a new drug lowers blood pressure compared to a placebo. Use α = 0.05, two-tailed.
#Find SE, t_cal, t_critical, p-value. Do you reject H₀?
n1=50
x1=130
s1=8
n2=50
x2=135
s2=9
#h0: m1=m2
#h1: m1/=m2
se=np.sqrt((s1**2/n1)+(s2**2/n2))
tcal=(x1-x2)/se
print("T Calculated:", tcal)
tcrit=stats.t.ppf(1-0.05/2, n1+n2-2) #two tailed
print("T Critical: +- ", tcrit)
p=(1-stats.t.cdf(abs(tcal),n1+n2-2))
print("P-value:  ", p)
if(p<0.05):
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import statsmodels.formula.api as smf
import statsmodels.api as sm

df=pd.read_csv('Advertising.csv', usecols=[1,2,3,4])
df.head()
df.describe()
df.corr()
```

```python
sns.regplot(data=df, x='TV', y='sales', order=1)
```

```python
lm_fit1=smf.ols('sales~TV', df).fit()
lm_fit1.summary()
```

```python
lm_fit_full=smf.ols('sales~TV + radio + newspaper', df).fit()
lm_fit_full.summary()
```

```python
lm_fit2=smf.ols('sales~TV + radio', df).fit()
lm_fit2.summary()
```

```python
# What if TV and radio together create extra sales beyond their individual effects?
lm_interact = smf.ols('sales ~ TV * radio', df).fit()
# This expands to: sales = α + β1*TV + β2*radio + β3*(TV×radio)
print(lm_interact.summary())
```

```python
lm = smf.ols('sales ~ TV * radio', df).fit()
# Predict sales for existing data
y_predict = lm.predict(df[['TV', 'radio']])
residuals = df['sales'] - y_predict
print(residuals.sum())
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.formula.api as sml
import statsmodels.api as sm

import pandas as pd, numpy as np
np.random.seed(42)
n = 100
df = pd.DataFrame({
  'hours': np.random.randint(1,10,n),
  'attendance': np.random.randint(50,100,n),
  'sleep': np.random.randint(4,9,n),
  'score': None})
df['score'] = (5*df['hours'] + 0.3*df['attendance']
              + np.random.normal(0,5,n)).clip(0,100).round(1)
df.corr() #target=score
```

```python
sns.regplot(df, x="hours", y="score", order=1)
```

```python
sns.pairplot(df[['hours', "attendance","sleep","score"]])
```

```python
lm_fit2=sml.ols("score ~ hours + attendance ", df).fit()
lm_fit2.summary()
```

```python
y_predict=lm_fit2.predict(df[['hours', 'attendance']])
(df['score']-y_predict).sum()
```

```python
b = lm_fit2.params
pred = b['Intercept'] + b['hours']*7 + b['attendance']*85
print(f"Predicted score: {pred:.2f}")
```
