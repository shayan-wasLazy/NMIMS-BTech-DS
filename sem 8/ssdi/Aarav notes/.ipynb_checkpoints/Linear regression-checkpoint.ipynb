```python
lm_fit1 = smf.ols("Sales~TV",df).fit()
lm_fit1.summary()
```

```python
lm_fit2 = smf.ols("Sales~Radio",df).fit()
lm_fit2.summary()
```

```python
lm_fit3 = smf.ols("Sales~Newspaper",df).fit()
lm_fit3.summary()
```

```python
lm_fit4 = smf.ols("Sales~TV + Radio + Newspaper",df).fit()
lm_fit4.summary()
```

```python
lm_fit5 = smf.ols("Sales~TV + Radio",df).fit()
lm_fit5.summary()
```

```python
#Best Model
lm_fit6 = smf.ols("Sales~TV * Radio",df).fit()
lm_fit6.summary()
```

```python
df["TV2"] = df["TV"].values**2
```

```python
lm_fit7 = smf.ols("Sales~ TV + TV2",df).fit()
lm_fit7.summary()
```

```python
df["Radio2"] = df["Radio"].values**2
```

```python
lm_fit8 = smf.ols("Sales~ Radio + Radio2",df).fit()
lm_fit8.summary()
```

```python
sns.regplot(data=df,x="Radio2",y="Sales",order=1)
plt.show()
```

```python
df["Newspaper2"] = df["Newspaper"].values**2
```

```python
lm_fit8 = smf.ols("Sales~ Newspaper + Newspaper2",df).fit()
lm_fit8.summary()
```
