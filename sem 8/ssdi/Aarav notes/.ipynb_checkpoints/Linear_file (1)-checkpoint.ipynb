```python
df = pd.read_csv(r"C:\Users\Aarav\OneDrive\Desktop\SSDI\DATA\Advertising.csv", usecols=[1,2,3,4])
```

```python
df.corr()
```

```python
#Best Model
lm_fit6 = smf.ols("Sales~TV * Radio",df).fit()
lm_fit6.summary()
```

```python
df["TV2"] = df["TV"].values**2
```

```python
lm_fit7 = smf.ols("Sales~TV + TV2",df).fit()
lm_fit7.summary()
```

```python
xdf = df[["TV","Radio"]]
```
