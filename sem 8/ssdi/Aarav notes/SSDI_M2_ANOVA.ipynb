```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
fit1=ols('horsepower~ origin*cylinders',data=df).fit()
annova1=sm.stats.anova_lm(fit1,typ=2)
annova1
```

```python
df['origin'] = df['origin'].replace({

     1: 'US',

     2: 'Europe',

     3: 'Japan'

 })
```

```python
df
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
from statsmodels.formula.api import ols
from statsmodels.base.model import Model
tukey=pairwise_tukeyhsd(df["horsepower"],groups=df["origin"]+df["cylinders"].astype(str))
tukey._results_table
```

```python
import pandas as pd
tukey_df = pd.DataFrame(tukey._results_table.data[1:], columns=tukey._results_table.data[0])
true_values = tukey_df[tukey_df['reject'] == True]
highest = true_values.loc[true_values['meandiff'].astype(float).idxmax()]
print(highest)
```

```python
import pandas as pd

df = pd.read_csv("HousingData.csv")
df = df.dropna()
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
fit1=ols('CRIM ~ RAD',data=df).fit()
annova1=sm.stats.anova_lm(fit1,typ=1)
annova1
```

```python
df["RAD"].unique()
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
from statsmodels.formula.api import ols
from statsmodels.base.model import Model
tukey=pairwise_tukeyhsd(df["CRIM"],groups=df["RAD"])
tukey._results_table
```

```python

C = df[df['RAD'] == 1]['CRIM']
R = df[df['RAD'] == 24]['CRIM']
```

```python
from scipy.stats import ttest_ind
import numpy as np
from scipy import stats
print(stats.ttest_ind(C,R,equal_var=True,alternative="greater"))
```

```python
WHEN RAD =1 CRIM IS LESS,WHCIH MEANS FARTHER FROM HIGHWAY MORE THE CRIME.
```
