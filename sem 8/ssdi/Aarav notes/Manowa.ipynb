```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant
```

```python
X = ab[["TV","Radio","Newspaper"]]
X = add_constant(X)
vif_ab = pd.DataFrame()
vif_ab ["Feature"] = X.columns

vif_ab ["VIF"] = [variance_inflation_factor(X.values,i) for i in range(len(X.columns))]

print(vif_ab)
```

```python
le = LabelEncoder()
for col in ['Gender', 'Student', 'Married', 'Ethnicity']:
    df1[col] = le.fit_transform(df1[col])
    
X = df1[['Income', 'Limit' , 'Cards', 'Age', 'Education','Gender', 'Student', 'Married', 'Ethnicity']]
X = add_constant(X)
vif_ab = pd.DataFrame()
vif_ab ["Feature"] = X.columns

vif_ab ["VIF"] = [variance_inflation_factor(X.values,i) for i in range(len(X.columns))]

print(vif_ab)
```
