```python
import numpy as np
import pandas as pd

np.random.seed(42)

n = 150

experience = np.random.randint(1,15,n)
education = np.random.randint(1,4,n)
projects = np.random.randint(1,20,n)

salary = (
    5000 +
    experience*2500 +
    education*12000 +
    projects*1500 +
    np.random.randn(n)*10000
)

df = pd.DataFrame({
    'Experience':experience,
    'Education':education,
    'Projects':projects,
    'Salary':salary
})
```

```python
df
```

```python
import seaborn as sns
```

```python
sns.regplot(x = df["Experience"], y = df["Salary"])
```
