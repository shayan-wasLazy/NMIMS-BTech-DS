```python
import pandas as pd
df=pd.read_csv("BostonHousing.csv")
df=df.dropna()
df
```

```python
df.info()
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
fit1 = ols('medv ~ chas',data=df).fit()
anova1 = sm.stats.anova_lm(fit1,typ=1)
anova1
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
tukey=pairwise_tukeyhsd(df["medv"],groups=df["chas"])
tukey._results_table
```

```python
a=df[df['chas']==1].medv
b=df[df['chas']==0].medv
```

```python
from scipy import stats
print(stats.ttest_ind(a,b,equal_var=True,alternative="greater"))

#river income greater than nonriver
```
