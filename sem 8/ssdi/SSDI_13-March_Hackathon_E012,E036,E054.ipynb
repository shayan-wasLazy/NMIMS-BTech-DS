```python
import pandas as pd
df=pd.read_excel('bank_case_study.xlsx')
df #balance call the premium customer based on balance
```

```python
df=df.dropna()
```

```python
df.isnull().sum()
```

```python
#job,  marital, education, default, housing, loan, cellular, campaign, previous, y
```

```python
import statsmodels.api as sm #balance and marital
from statsmodels.formula.api import ols
fit=ols('balance ~ C(job) + C(marital) + C(education) + C(default) + C(housing) + C(loan) + C(contact) + campaign + previous + C(y)', data=df).fit()  
anova=sm.stats.anova_lm(fit, typ=1)
anova #job, education, default, loan, marital
```

```python
import statsmodels.api as sm # balance vs Job
from statsmodels.formula.api import ols
fit=ols('balance ~ job*education', data=df).fit()  
anova=sm.stats.anova_lm(fit, typ=1)
anova #significantly affects
```

```python
import statsmodels.api as sm #balance and marital
from statsmodels.formula.api import ols
fit=ols('balance ~ job*marital*education', data=df).fit()  
anova=sm.stats.anova_lm(fit, typ=1)
anova #doesn't significantly affects
```

```python
import statsmodels.api as sm #balance and marital
from statsmodels.formula.api import ols
fit=ols('balance ~ job*marital*education*loan', data=df).fit()  
anova=sm.stats.anova_lm(fit, typ=1)
anova #doesn't significantly affects
```

```python
import statsmodels.api as sm #balance and marital
from statsmodels.formula.api import ols
fit=ols('balance ~ job*marital*education*default*loan', data=df).fit()  
anova=sm.stats.anova_lm(fit, typ=1)
anova #doesn't significantly affects
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
tukey=pairwise_tukeyhsd(df["balance"],groups=df["education"])
tukey._results_table
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
tukey=pairwise_tukeyhsd(df["balance"],groups=df["loan"])
tukey._results_table
```

```python
a = df[(df["education"] == 'primary') & (df["loan"] == "yes")][["balance"]]
b = df[(df["education"] == 'unknown') & (df["loan"] == "yes")][["balance"]]
```

```python
from scipy import stats
print(stats.ttest_ind(a, b, equal_var=True, alternative="greater"))
```

```python

d=df[(df['y']=='no') & (df["education"] == 'unknown') & (df["loan"] == "yes") & (df['default']=='no')]
```

The presence of high-balance customers within the (Unknown Education + Loan group) suggests that balance levels are not solely determined by loan status. This indicates that the relationship between balance and loan status is complex and cannot be explained by a simple linear trend.
