```python
import pandas as pd
df=pd.read_csv('iris.csv')
df
```

```python
from statsmodels.multivariate.manova import MANOVA
df.columns = ['sepal_length','sepal_width','petal_length','petal_width','variety']
maov = MANOVA.from_formula(
    'sepal_length + sepal_width + petal_length + petal_width ~ variety',
    data=df
)
print(maov.mv_test())
```

```python
import statsmodels.api as sm 
from statsmodels.formula.api import ols
fit=ols('sepal_length ~ variety', data=df).fit()  
anova=sm.stats.anova_lm(fit, typ=1)
anova 
```

```python
import statsmodels.api as sm 
from statsmodels.formula.api import ols
fit=ols('sepal_width ~ variety', data=df).fit()  
anova=sm.stats.anova_lm(fit, typ=1)
anova 
```

```python
import statsmodels.api as sm 
from statsmodels.formula.api import ols
fit=ols('petal_width ~ variety', data=df).fit()  
anova=sm.stats.anova_lm(fit, typ=1)
anova 
```

```python
import statsmodels.api as sm 
from statsmodels.formula.api import ols
fit=ols('petal_length ~ variety', data=df).fit()  
anova=sm.stats.anova_lm(fit, typ=1)
anova 
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
tukey=pairwise_tukeyhsd(df["petal_length"],groups=df["variety"])
tukey._results_table
```
