```python
from scipy import stats
import numpy as np

# H0: p1 - p2 = 0
# H1: p1 - p2 != 0  (two-tailed)

n1 = 400; p1 = 0.415
n2 = 380; p2 = 0.539
alpha = 0.05

# pooled proportion (since H0 assumes p1 = p2)
p_pool = (n1*p1 + n2*p2) / (n1 + n2)
q_pool = 1 - p_pool

# SE using pooled proportion
se = np.sqrt(p_pool * q_pool * (1/n1 + 1/n2))

# z statistic (not t — proportions use normal distribution)
zcal = (p1 - p2) / se

# critical value and p-value
zcrit = stats.norm.ppf(1 - alpha/2)
p_value = 2 * (1 - stats.norm.cdf(abs(zcal)))

print("Pooled p:    ", round(p_pool, 4))
print("SE:          ", round(se, 4))
print("Z-Calculated:", round(zcal, 4))
print("Z-Critical: ±", round(zcrit, 4))
print("P-value:     ", round(p_value, 4))

if abs(zcal) > zcrit:
    print("Reject H₀ — proportions are significantly different")
else:
    print("Fail to reject H₀")
```
