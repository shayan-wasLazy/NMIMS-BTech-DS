```python
import numpy as np
from scipy import stats
from statistics import stdev
```

```python
n1=60
n2=75
x1=86
s1=6
x2=82
s2=9
alpha=0.01
se=np.sqrt((s1**2/n1)+(s2**2/n2))
print(se)
tcal=(x1-x2)/se
print(tcal)
tpos=stats.t.ppf(1-alpha/2, n1+n2-2)
tneg=stats.t.ppf(alpha/2, n1+n2-2)
print(tneg, "    ", tpos)
p=stats.t.cdf(tcal,n1+n2-2)
print(1-p)  #reject the hypothesis
```

```python
def ttest_two_sample(data1, data2, alt='two', alpha=0.05):  # fix 1+extra
    mean1=np.mean(data1); mean2=np.mean(data2)  # fix 1: use parameters
    std1=stdev(data1);   std2=stdev(data2)
    n1=len(data1);       n2=len(data2)
    df=n1+n2-2
    se=np.sqrt((std1**2/n1)+(std2**2/n2))
    tcal=(mean1-mean2)/se
    if alt=='two':
        tcrit=stats.t.ppf(1-alpha/2, df)
        p=2*(1-stats.t.cdf(abs(tcal), df))  # fix 2: multiply by 2
        reject=abs(tcal)>tcrit
        return {'tcal':tcal, 'tcrit':(- tcrit, tcrit), 'p':p, 'reject':reject}
    elif alt=='right':
        tcrit=stats.t.ppf(1-alpha, df)
        p=1-stats.t.cdf(tcal, df)
        reject=tcal>tcrit
        return {'tcal':tcal, 'tcrit':tcrit, 'p':p, 'reject':reject}
    elif alt=='left':  # fix 3: explicit left-tail branch
        tcrit=stats.t.ppf(alpha, df)
        p=stats.t.cdf(tcal, df)
        reject=tcal<tcrit
        return {'tcal':tcal, 'tcrit':tcrit, 'p':p, 'reject':reject}
result=[]
result=twosided(a, b,"two")
print(result)
```
