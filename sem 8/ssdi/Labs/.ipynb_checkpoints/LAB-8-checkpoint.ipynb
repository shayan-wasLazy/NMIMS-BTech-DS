```python
import pandas as pd
df = {"Drink1": [2,3,7,2,6],"Drink2":[10,8,7,5,10],"Drink3":[10,13,14,13,15]}
df=pd.DataFrame(data=df)
df
import statsmodels.api as sm
```

```python
df1 = pd.melt(df.reset_index(),id_vars = ["index"],value_vars=["Drink1","Drink2","Drink3"])
df1.columns=["index","Treatments","Value"]
print(df1)
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

fit1 = ols('Value ~ Treatments',data=df1).fit()   ##Value is X and Treatments is Factor variable
anova1 = sm.stats.anova_lm(fit1,typ=1)
anova1
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
#from statsmodels.formula.api import ols
#from statsmodels.base.model import Model
#from statsmodels.stats.multicomp import pairwise_tukeyhsd
tukey = pairwise_tukeyhsd(df1["Value"],groups=df1["Treatments"])
tukey._results_table
```

```python
import pandas as pd
import numpy as np
from statsmodels.formula.api import ols
import statsmodels.stats.multicomp as mc
import statsmodels.api as sm
df = pd.DataFrame({'water': np.repeat(['daily', 'weekly'], 15),
                   'sun': np.tile(np.repeat(['low', 'med', 'high'], 5), 2),
                   'height': [6, 6, 6, 5, 6, 5, 5, 6, 4, 5,
                              6, 6, 7, 8, 7, 3, 4, 4, 4, 5,
                              4, 4, 4, 4, 4, 5, 6, 6, 7, 8]})
df
```

```python
fit2 = ols('height ~ C(water) + C(sun) + C(water):C(sun)', data=df).fit()
anova2 = sm.stats.anova_lm(fit2,typ=2)
print(anova2)
```

```python

import statsmodels.stats.multicomp as mc
```

```python
com1=mc.MultiComparison(df['height'],df["sun"])
test1=com1.tukeyhsd()
test1.summary()
```

```python
com2=mc.MultiComparison(df['height'],df["water"])
test2=com2.tukeyhsd()
test2.summary()
```

```python
import statsmodels.stats.multicomp as mc
com=mc.MultiComparison(df['height'],df["sun"]+df['water'])
test=com.tukeyhsd()
test.summary()
```
