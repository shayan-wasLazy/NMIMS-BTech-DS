👉Testing of Hypothesis

```python
##We have a training program designed to upgrade the supervisory skills of 
#production-line supervisors.Because the program is self-administered,supervisors
#require diffrent numbers of hours to complete the program. A study of past participants
#indicates that the mean length of time spent pon the program is 500 hours and this nomally
#distibuted random variable has standard deviation of 100 hours.
```

```python
#Q1. what is probability that a participants selected at random will require more than
#500 hours to complete the program?
import numpy as np
from scipy import stats
Zcal1 = (500-500)/100
print(stats.norm.cdf(Zcal1))

#Q2. what is probability that a participants selected at random will take between 500 and
#650 hours to complete the program?
Zcal2 = (650-500)/100
print(stats.norm.cdf(Zcal2)-0.5)

#Q3.what is probability that a participants selected at random will take more than 700
#hours to complete the program?
Zcal3 = (700-500)/100
print(1-stats.norm.cdf(Zcal3))

##Q4.what is probability that a participants selected at random will take btween 550 to 650
#hours to complete the program?

Zcalh = (650-500)/100
Zcall = (550-500)/100
print(stats.norm.cdf(Zcalh)-stats.norm.cdf(Zcall))
```

```python
##Two side tabular value
org_alpha=0.05
alpha=0.025      ##alpha=org_alpha/2
Z_cal_positive=stats.norm.ppf(1-alpha,0,1)
Z_cal_negative=stats.norm.ppf(alpha,0,1)
print(Z_cal_positive)
print(Z_cal_negative)
##one sided 5%
print()
print(stats.norm.ppf(1-alpha*2),0,1)    ##org_alpha=alpha*2
print(stats.norm.ppf(alpha*2),0,1) 
print()

n=21
df=20    ##df=n-1
t_cal_negative=stats.t.ppf(alpha,df) ##accept H0   ##Two tail 5% error and one tail 2.5% error
t_cal_positive=stats.t.ppf(1-alpha,df)   
print(t_cal_negative)
print(t_cal_positive)

##one sided 5%
print()
print(stats.t.ppf(alpha*2,df))

##find out p value from tabular value
print()
print(stats.norm.cdf(Z_cal_positive))
print()
print(stats.t.cdf(t_cal_positive,df))
```

```python
#American Theaters know that a certain hit movie ran an average of 84 days in each city, 
#and the corresponding deviation is 10 days. The manager of the south-eastern district 
#is interested in comparing the movies popularity in his region with that in all of American’s 
#other theaters. He randomly chose 75 theaters in his region and found that they 
#ran the movie on an average of 81.5 days. State the appropriate hypotheses for testing 
#whether there was a significant difference in the length of the picture’s run between 
#theaters in the south-eastern district and all of American’s other theaters at 1% level.
```

```python
##Testing of hypothesis for mean
##The null hypothesis is that μ = 84. We begin with computing the test statistic
##Z test
import numpy as np
xbar = 81.5           # sample mean 
μ =    84        # Null hypothesized value
#μ =/= 84       ##Alternative hypothesized value
sigma = 10           # population standard deviation 
n = 75               # sample size
alpha=0.01
se = sigma/np.sqrt(n)
Z_cal = (xbar-μ)/se
print(Z_cal)
```

```python
from scipy import stats
alpha=0.01
stats.norm.ppf(1-alpha)   ##Two sided
z_table_positive=(stats.norm.ppf(1-alpha/2))
z_table_negative=-(stats.norm.ppf(1-alpha/2))
print(Z_cal)
print(z_table_positive, "     " ,z_table_negative)
##Accept Ho
print()
##or

print(stats.norm.cdf(Z_cal))  
## p value >0.01that's why Accept Ho
##0r
print()
print(μ+z_table_negative*se,'      ', μ+z_table_positive*se)
##Sample mean i.e. 81.5 is inside the range that's why accept Ho
```

```python
#Hinton Press hypothesizes that the average life of its largest web press is 14,500 hours. 
#They know that the standard deviation of press life is 2,100 hours. From a sample of 
#25 presses, the company finds a sample mean of 13,000 hours. At a 0.01 significance level,
#should the company conclude that the average life of the presses is less than 
#the hypothesized 14,500 hours?
```

```python
##Testing of hypothesis for mean
##The null hypothesis is that μ ≥ 14500. We begin with computing the test statistic
##Z test
import numpy as np
xbar = 13000           # sample mean 
μ =  14500        # Null hypothesized value
#μ < 14500      ##Alternative hypothesized value
sigma = 2100           # population standard deviation 
n = 25               # sample size
alpha=0.01
se = sigma/np.sqrt(n)
Z_cal = (xbar-μ)/se
print(Z_cal)
```

```python
alpha=0.01
stats.norm.ppf(1-alpha)   ##Two sided
z_table_negative=-(stats.norm.ppf(1-alpha))
print(Z_cal)
print(z_table_negative, "     " ,'3')
##Reject Ho
print()
##or

print(stats.norm.cdf(Z_cal))  
## p value <0.01that's why Reject Ho
##0r
print()
print(μ+(z_table_negative*se),'      ', 'inf')
##Sample mean i.e. 1300 is outside the range that's why Reject Ho
```

```python
#Investigators wanted to determine whether children who were born very prematurely 
#have poorer cognitive ability than children born at full term. To investigate they enrolled 
#100 children who had been born prematurely and measured their IQ in order to compare 
#the sample mean to a normative IQ of 100. In essence, the normative IQ is a historical 
#or external (μ0). In the sample of 100 children who had been born prematurely the data
#for IQ were as follows: X̄ = 95.8, SD = 17.5" . 
#The investigators wish to use a one-tailed test of significance.
```

```python
##t test
import numpy as np
xbar = 95.8            # sample mean 
μ = 100           # Null hypothesized value
#μ < 100       ##Alternative hypothesized value
sample_sd = 17.5            # sample standard deviation 
n = 100                 # sample size
se = sample_sd/np.sqrt(n)
t_cal = (xbar-μ)/se
print(t_cal)
```

```python
alpha=0.05
#stats.t.ppf(1-alpha,(n-1))  ##two sided
#t_table_positive=(stats.t.ppf(1-alpha,(n-1)))
t_table_negative=-(stats.t.ppf(1-alpha,(n-1)))
#print(t_table_positive, "     " ,t_table_negative)
print(t_cal)
print(t_table_negative, "     " ,3)
##Reject Ho
print()
##or
print(stats.t.cdf(t_cal,(n-1)))  
## p value <0.05 that's why Reject Ho
print()
print()
print(μ+(t_table_negative*se),'      ', 'inf')
##Sample mean i.e. 100 is outside the range that's why Reject Ho
```

```python
from scipy import stats
import numpy as np
from statistics import stdev
X = [10,12,13,14,15,2,7,8]
df=len(X)-1
xbar = np.mean(X)            # sample mean 
μ = 12          # Null hypothesized value
#μ < 100       ##Alternative hypothesized value
sample_sd = stdev(X)           # sample standard deviation 
n = len(X)                 # sample size
se = sample_sd/np.sqrt(n)
t_cal = (xbar-μ)/se
print(t_cal)

alpha=0.01/2
stats.t.ppf(1-alpha,(n-1))  ##two sided
t_table_positive=(stats.t.ppf(1-alpha,(n-1)))
t_table_negative=-(stats.t.ppf(1-alpha,(n-1)))
print(t_table_positive, "     " ,t_table_negative)
##Accept HO
```

```python

##Testing of Hypothesis through pre_trained library two sided
##mu:H0=12
##mu:Ha=/=12
X = np.array([10,12,13,14,15,2,7,8])
df=len(X)-1
print(stats.ttest_1samp(X,popmean=12,alternative='two-sided'))  ##{‘two-sided’, ‘less’, ‘greater’}  ##accept Ho
##or
print()
print(stats.t.cdf(-1.2264982650902285,df)*2)
print()
print(stats.t.ppf(0.025,df))  ##accept H0   ##Two tail 5% error and one tail 2.5% error
print(stats.t.ppf(0.975,df))
```

```python
##Testing of Hypothesis through pre_trained library one sided
##mu:H0≥12
##mu:Ha<12
X = np.array([10,12,13,14,15,2,7,8])
print(stats.ttest_1samp(X,popmean=12,alternative='less'))  ##{‘two-sided’, ‘less’, ‘greater’}  ##accept Ho
##or
print()
print(stats.t.cdf(-1.2264982650902285,df))
print()
print(stats.t.ppf(0.05,df))  ##accept H0  ##Two tail 10% error and one tail 5% error
```
