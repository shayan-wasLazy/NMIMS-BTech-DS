Maximum likelihood estimation Poisson Distribution

```python
from scipy.stats import poisson
import numpy as np
import matplotlib.pyplot as plt
```

```python
mean=5
print(poisson.pmf(k=2,mu=mean)) ##Likelihood
print(poisson.cdf(k=2,mu=mean))
x = np.random.poisson(5,10000)
plt.hist(x)
plt.show()
```

```python
#X=np.arange(0, 10)
X=np.random.poisson(lam=5,size=1000)
def pl(lam,X):
    
    log=[]       
    for i in range(0,len(X)):
    
        logs = np.log(poisson.pmf(k=X[i],mu = lam))
        log.append(logs)
    return np.sum(log)

```

```python
pl(5,X)
```

```python
lams = np.arange(1,10,1)

# likelihood
ls = []
for lam in lams:
    ls1 = (pl(lam,X))
    ls.append(ls1)

lams[np.argmax(ls)]
plt.plot(lams,ls)
```

```python
from scipy.stats import bernoulli
from scipy.optimize import minimize
import numpy as np
```

```python
x = np.random.uniform(0,1,10)
up=0.1
x=(x>up)*1
print(x)
p=0.1
bernoulli.pmf(x,p)
```

```python
up = 0.8
x = np.random.uniform(0,1,1000)
x = (x<up)*1.0

p=0.1

def log_h(p,x):
    l = np.sum(np.log(bernoulli.pmf(x,p)))
    return -l
def con(p):
    return p
const={'type':'ineq','fun':con}
print(minimize(log_h,p,args=(x,),constraints=const))
```
