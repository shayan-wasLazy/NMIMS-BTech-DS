MLE of Binomial distribution 

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# 1. DATA: 20 startups, 6 successes  p=6/20=0.30
n = 20
k = 6

# 2. Test different possible values of p (from 0 to 1)
p_range = np.linspace(0, 1, 100)
likelihood = binom.pmf(k, n, p_range)

# 3. Find the MLE (the p that maximizes likelihood)
mle_p = p_range[np.argmax(likelihood)]

# 4. PLOT THE RESULTS
plt.figure(figsize=(10, 6))
plt.plot(p_range, likelihood, label='Likelihood Function', color='blue', lw=2)
plt.axvline(mle_p, color='red', linestyle='--', label=f'MLE (p={mle_p:.2f})')
plt.title(f"Binomial MLE: Finding the 'Most Likely' Success Rate")
plt.xlabel("Probability of Success (p)")
plt.ylabel("Likelihood")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print(f"The most likely success rate for our startups is: {mle_p:.2f}")
```

```python
import numpy as np
from scipy.stats import binom
from scipy.optimize import minimize
from scipy.special import factorial
```

```python
n=100
k=40
inta=[0.1]
def BLL(p,n,k):
    if p<0 or p>1:
        return 1e05
    a1=factorial(n)/((factorial(n-k)*factorial(k)))
    a2=(p**k)*((1-p)**(n-k))   ##np.power()
    LL=a1*a2
    logLL=np.log(LL)
    return -logLL
results=minimize(BLL,inta,args=(n,k),bounds=[(5e-5,1-5e-5)])
results.x
```
