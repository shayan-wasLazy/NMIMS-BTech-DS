Q.A large hotel chain is trying to decide whether to convert more of its rooms to non-smoking rooms.
In a random sample of 400 guests last year, 166 had requested non-smoking rooms. This year, 205
guests in a sample of 380 preferred the non-smoking rooms. Would you recommend that the hotel
chain convert more rooms to non-smoking? Support your recommendation by testing the appropriate
hypotheses at a 0.01 level of significance.

```python
#H0:p1≥p2   p1-p2>=0
#H1:P1<P2
import numpy as np
n1 =400
n2 = 380
p1 = 166/400
p2 = 205/380
alpha =0.01
phat = (n1*p1+n2*p2)/(n1+n2)
se = np.sqrt((phat*(1-phat))*(1/n1+1/n2))   ##sqrt(phat*qhat/n1+phat*qhatq/n2)
zcal = ((p1-p2)-0)/se    ##0 because H0:p1-p2>=0
from scipy import stats
print(stats.norm.cdf(zcal))  ##Reject Ho
print(zcal)
```

```python
stats.norm.ppf(1-alpha)   ##Two sided
#z_table_positive=(stats.norm.ppf(1-alpha))
z_table_negative=-(stats.norm.ppf(1-alpha))
print(zcal)
print(z_table_negative, "     " ,"inf")
##Reject Ho
print()
##or
print(stats.norm.cdf(zcal))  
## p value <0.01that's why Reject Ho
##0r
print()
print(0+z_table_negative*se,'      ', 3)
##Sample mean i.e. p1-p2=-0.1245 is outside the range that's why reject Ho
```

```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest
count = np.array([166, 205])
nobs = np.array([400, 380])
stat, pval = proportions_ztest(count, nobs)
print('{0:0.3f}'.format(pval))
print('{0:0.3f}'.format(stat))
proportions_ztest(count, nobs)
```

```python
import numpy as np
from scipy import stats
from statistics import stdev
a = np.array([56, 128, 12, 123, 64, 78, 763])
b = np.array([46,100,5,121,54,80,700])
c = a-b
xbar = np.mean(c)           # sample mean 
μ = 0          # Null hypothesized value
#μ =/= 0       ##Alternative hypothesized value
sample_sd = stdev(c)            # sample standard deviation 
n = len(c)                 # sample size
se = sample_sd/np.sqrt(n)
t_cal = (xbar-μ)/se
print(t_cal)
print((1-(stats.t.cdf(t_cal,n-1)))*2)
```

```python
##Through pre trained library
print(stats.ttest_rel(a,b))
```
