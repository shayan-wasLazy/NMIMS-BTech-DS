Connie Rodrigues, the Dean of Students at Midstate College, is wondering about grade 
distributions at the school. She has heard grumblings that the GPAs in the Business School 
are about 0.25 lower than those in the College of Arts and Sciences. A quick random 
sampling produced the following GPAs.

Business:2.86,2.77,3.18,2.80,3.14,2.87,3.19,3.24,2.91,3,2.83

Arts and Science: 3.35,3.32,3.36,3.63,3.41,3.37,3.45,3.43,3.44,3.17,3.26,3.18,3.41
    
Do thests data indicates that there is a factual basis for the grumblings?State and test
appropriate hypothesis at alpha=0.02

```python
import numpy as np
from scipy import stats
Bus=np.array([2.86,2.77,3.18,2.80,3.14,2.87,3.19,3.24,2.91,3,2.83])
As = np.array([3.35,3.32,3.36,3.63,3.41,3.37,3.45,3.43,3.44,3.17,3.26,3.18,3.41])
##The null hypothesis is that μ1-μ1 =  0 or μ1 ≥ μ1 . We begin with computing the test statistic
#Ho:μB-μA =  -0.25          # hypothesized value
#Ha:μB-μA =/= -0.25

xbarB = np.mean(Bus)           # sample means 
xbarA = np.mean(As)


s1 = stdev(Bus)           # sample standard deviation 
s2 = stdev(As)
n1 = len(Bus)               # sample size
n2 = len(As)
sp = np.sqrt(((n1-1)*s1**2+(n2-1)*s2**2)/(n1+n2-2))

se = sp*np.sqrt(1/n1+1/n2)
t_cal = ((xbarB-xbarA)-(-0.25))/se
print()
print(t_cal)

##one side 10%
alpha=0.01
df=n1+n2-2

```

```python
df=n1+n2-2
alpha=0.02
stats.t.ppf(1-alpha,df)   ##Two sided
t_table_positive=(stats.t.ppf(1-alpha,df))
t_table_negative=-(stats.t.ppf(1-alpha,df))
print(t_cal)
print(t_table_positive, "     " ,t_table_negative)
##Accept Ho
print()
##or

print(stats.t.cdf(t_cal,df))  
## p value >0.01that's why Accept Ho
##0r
print()
print(-0.25+t_table_negative*se,'      ', -0.25+t_table_positive*se)
##Sample mean i.e. 81.5 is inside the range that's why accept Ho
```

A consumer-research organization routinely selects several car models each year 
and evaluates its fuel efficiency. In this year's study of two similar subcompact models 
from two different automakers were tested. The average gas mileage for 12 cars of 
brand A was 27.2 miles per gallon, with a standard deviation of 3.8 mpg. The nine brand 
B cars that were tested averaged 32.1 mpg, with a standard deviation of 4.3 mpg. At alpha=0.01,
should it conculde that brand A cars have lower average gas mileage than do brand B cars?

```python
import numpy as np
from scipy import stats
##The null hypothesis is that μ1-μ1 ≥  0 or μ1 ≥ μ1 . We begin with computing the test statistic
#Ho:μ1-μ1 ≥  0            # hypothesized value
#Ha:μ1-μ1 < 0

xbar1 = 27.2           # sample means 
xbar2 = 32.1


s1 = 3.8           # sample standard deviation 
s2 = 4.3
n1 = 12               # sample size
n2 = 9
sp = np.sqrt(((n1-1)*s1**2+(n2-1)*s2**2)/(n1+n2-2))

se = sp*np.sqrt(1/n1+1/n2)
t_cal = ((xbar1-xbar2)-0)/se
print()
print(t_cal)

##one side 10%
alpha=0.01
df=n1+n2-2

```

```python
alpha=0.01
df=n1+n2-2
#stats.t.ppf(1-alpha,(n-1))  ##two sided
#t_table_positive=(stats.t.ppf(1-alpha,df))
t_table_negative=-(stats.t.ppf(1-alpha,df))
#print(t_table_positive, "     " ,t_table_negative)
print(t_cal)
print(t_table_negative, "     " ,3)
##Reject Ho
print()
##or
print(stats.t.cdf(t_cal,df))
## p value <0.05 that's why Reject Ho
print()
print()
print(0+(t_table_negative*se),'      ', 'inf')
##Sample mean differnece i.e. -4.9 is outside the range that's why Reject Ho
```

```python
##Testing of Hypothesis through pre_trained library one sided
alpha=0.05
a = np.array([56, 128.6, 12, 123.8, 64.34, 78, 763.3])
b = np.array([1.1, 2.9, 4.2])
print(stats.ttest_ind(a,b,equal_var=True,alternative="two-sided"))
print(stats.ttest_ind(a,b,equal_var=False))
df =(len(a)+len(b))-2
t_table_negative=stats.t.ppf(alpha/2,df) ##accept H0   ##Two tail 5% error and one tail 2.5% error
t_table_positive=stats.t.ppf(1-(alpha/2),df)
print()

print(t_table_negative,'    ',t_table_positive)     ##Accept Ho
```
