```python
##Logistic Regression
import pandas as pd
import numpy as np
from sklearn import preprocessing
import matplotlib.pyplot as plt 
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
import seaborn as sns
import statsmodels.formula.api as smf
```

```python
df=pd.read_csv('chd.csv',header=0).dropna()
```

```python
df.head(1)
df.isna().sum()
```

```python
df.info()
```

```python
fit=smf.logit('CHD~Age',df).fit()
fit.summary()
```

```python
lr=LogisticRegression()
X=np.array(df['Age']).reshape(100,1)   ##df.iloc,df['spl','s
Y=df['CHD']
lr_fit=lr.fit(X,Y)
print(lr_fit.coef_)
print(lr_fit.intercept_)
```

```python
Y_pred=fit.predict(df['Age'])
Y_pred=np.round(Y_pred)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(Y, Y_pred)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(Y, Y_pred)
```

```python
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
model = LogisticRegression()
K=5
kfold = KFold(K,random_state=0,shuffle=True)
mse_cv = cross_val_score(model,X,Y,cv=kfold,scoring='accuracy')
mse_cv
```

```python
X=df['Age']
Y=df['CHD']
Xtrain,Xtest,Ytrain,Ytest=train_test_split(X,Y,test_size=0.20,random_state=123)
```

```python
df_train=pd.concat([Xtrain,Ytrain],axis=1)
df_train.head(2)
```

```python
fit_train=smf.logit('CHD~Age',df_train).fit()
fit_train.summary().tables[1]
```

```python
Y_prd_train=fit_train.predict(df_train['Age'])
Y_prd_train=np.round(Y_prd_train)
print(confusion_matrix(df_train['CHD'],Y_prd_train))
print(accuracy_score(df_train['CHD'],Y_prd_train))
```

```python
Y_prd_test=fit_train.predict(Xtest)
Y_prd_test=np.round(Y_prd_test)
print(confusion_matrix(Ytest,Y_prd_test))
print(accuracy_score(Ytest,Y_prd_test))
```
