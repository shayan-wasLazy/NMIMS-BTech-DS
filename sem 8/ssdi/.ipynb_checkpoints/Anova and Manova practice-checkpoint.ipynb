```python
import pandas as pd
from statsmodels.formula.api import ols
import statsmodels.api as sm
from statsmodels.stats.multicomp import pairwise_tukeyhsd
```

```python
df=pd.read_csv("iris.csv")
print(df.columns)
df = df.rename(columns={
    'sepal.length': 'sepallength', 
    'sepal.width': 'sepalwidth',
    'petal.length': 'petallength',
    'petal.width': 'petalwidth'
})
fit=ols('sepallength ~ variety', data=df).fit()
anova=sm.stats.anova_lm(fit, type=1)
print(anova)
```

```python
tukey=pairwise_tukeyhsd(df['sepallength'], df['variety'])
print(tukey.summary())
```

```python
df=pd.read_excel('ANOVA.xlsx')
fit=ols("Method ~ Var", data=df).fit()
anova=sm.stats.anova_lm(fit, type=1)
print(anova)
```

```python
tukey=pairwise_tukeyhsd(df['Method'], df['Var'])
print(tukey._results_table)
```

```python
from statsmodels.multivariate.manova import MANOVA
df=pd.read_csv("iris.csv")
print(df.columns)
df = df.rename(columns={
    'sepal.length': 'sepallength', 
    'sepal.width': 'sepalwidth',
    'petal.length': 'petallength',
    'petal.width': 'petalwidth'
})
ma=MANOVA.from_formula(' sepallength + sepalwidth + petallength + petalwidth ~ variety', data=df)
print(ma.mv_test())
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Run for EACH numeric column separately
tukey1 = pairwise_tukeyhsd(df['petallength'], groups=df['variety'])
tukey2 = pairwise_tukeyhsd(df['petalwidth'], groups=df['variety'])
tukey3 = pairwise_tukeyhsd(df['sepallength'], groups=df['variety'])
tukey4 = pairwise_tukeyhsd(df['sepalwidth'], groups=df['variety'])
print(tukey1._results_table)
```
